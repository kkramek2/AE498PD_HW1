```python
import numpy as np
import sympy as sym
from scipy.optimize import minimize
```

#### (1) The Earth and asteroid 99942 Apophis

```python
# Given

# Earth
a_E = 1.4765067E8                     # km
e_E = 9.1669995E-3
i_E = np.deg2rad(4.2422693E-3)        # rad
omega_E = np.deg2rad(6.64375167E1)    # rad
Omega_E = np.deg2rad(1.4760836E1)     # rad

# Apophis
a_A = 1.3793939E8                       # km
e_A = 1.9097084E-1
i_A = np.deg2rad(3.3356539)             # rad
omega_A = np.deg2rad(1.2919949E+02)     # rad
Omega_A = np.deg2rad(2.0381969E+02)     # rad

# Define functions for rotation matrices
def rotation_matrix(i, omega, Omega):
    R3_Omega = np.array([[np.cos(Omega), -np.sin(Omega), 0],
                         [np.sin(Omega), np.cos(Omega), 0],
                         [0,            0,              1]])
    R1_i = np.array([[1, 0,         0],
                     [0, np.cos(i), -np.sin(i)],
                     [0, np.sin(i), np.cos(i)]])
    R3_omega = np.array([[np.cos(omega), -np.sin(omega), 0],
                        [np.sin(omega), np.cos(omega), 0],
                        [0,             0,              1]])
    
    return R3_Omega @ R1_i @ R3_omega

# Define function to calculate the state vector r
def position(a, e, i, omega, Omega, f):
    r = a*(1 - e**2)/(1 + e*np.cos(f))

    r_PQW = r * np.array([np.cos(f), np.sin(f), 0])

    Q = rotation_matrix(i, omega, Omega)

    return Q @ r_PQW

# Define function to calculate the orbital distance between two bodies
def orbital_dist(fs, body1, body2):
    f1, f2 = fs

    r1 = position(*body1, f1)
    r2 = position(*body2, f2)

    diff = np.linalg.norm(r1 - r2)

    return diff**2

# Organize body orbital elements into list
# Earth
body1 = (
    a_E,
    e_E,
    i_E,
    omega_E,
    Omega_E
)

# Apophis
body2 = (
    a_A,
    e_A,
    i_A,
    omega_A,
    Omega_A
)

# Range of true anomalies within the orbit
fs = np.linspace(0, 2*np.pi, 200)

# Define best guess and best value for the MOID, starting with infinity
best_guess = None
best_val = np.inf

# Update the MOID to get the smallest distance between the bodies
for f1 in fs:
    for f2 in fs:
        d = orbital_dist((f1, f2), body1, body2)
        if d < best_val:
            best_val = d
            best_guess = np.array([f1, f2])

# Refine results using scipy minimize
result = minimize(orbital_dist, best_guess, args=(body1, body2), bounds=[(0, 2*np.pi), (0, 2*np.pi)])

moid_sq = result.fun
moid = np.sqrt(moid_sq)

km_to_au = 1.496E8
moid_AU = moid/km_to_au

print("MOID calculations between Earth and 99942 Apophis:")

print("MOID (km) =", moid, "km")
print("MOID (AU) =", moid_AU, "AU")
# print("True anomaly on orbit 1 (rad) =", result.x[0])
# print("True anomaly on orbit 2 (rad) =", result.x[1])
print("True anomaly where MOID was found in form (f_1, f_2) =", "(",result.x[0],",", result.x[1], ")", "rad")
```
